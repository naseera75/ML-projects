# Ordinal Encoder

```python
import pandas as pd 
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('categorical_regression_dataset.csv')
```

```python
x=df.drop(columns='Salary')
y=df['Salary']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encoder=OrdinalEncoder(categories=[['Beginner','Intermediate','Advanced'],['High School','Diploma','Bachelor','Master'],['IT','HR','Marketing','Finance','Sales']],handle_unknown="use_encoded_value",unknown_value=-1)
encoder.fit_transform(xtrain[['Experience_Level','Education_Level','Department']])
# encoder.transform([['Ohh','No','Yss']])
```
