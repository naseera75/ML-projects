```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('advertising.csv')
df.info()
```

```python
df.corr()
```

```python
# plt.scatter(df.drop(columns='Sales'),df['Sales'])
```

```python
x=df.drop(columns='Sales')
y=df['Sales']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
poly=PolynomialFeatures(degree=3)

poly.fit_transform(xtrain) #In  transform fit we have to pass only xtrain values but not xtest values
poly.transform(xtest)
```

```python
xtrain_poly=poly.fit_transform(xtrain)
xtest_poly=poly.transform(xtest)
```

```python
model=LinearRegression()
model.fit(xtrain_poly,ytrain)
```

```python
model=LinearRegression()
model.fit(xtest_poly,ytest)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error,root_mean_squared_error
```

```python
y_pred_train=model.predict(xtrain_poly)
r2_score(ytrain,y_pred_train)
```

```python
y_pred_test=model.predict(xtest_poly)
r2_score(ytest,y_pred_test)
```

```python
model.score(xtrain_poly,ytrain)
```

```python
model.score(xtest_poly,ytest)
```
